```python
import pandas as pd, json, yaml, numpy as np
```

```python
load_file='B97-3_aug-pcseg-2_H2O/STORAGE_POLARIZ-FF_B97-3_aug-pcseg-2_H2O.yaml'
previous_data='../data/props_dfas_with_smaller_basis.json'

with open(load_file) as rd:
    storage=yaml.safe_load(rd)
old_df=pd.read_json(previous_data)
```

```python
the_record=old_df[ 
    (old_df['method']=='b97-3' )
    & ( old_df['basis']=='aug-pcseg-2')
    & ( old_df['molecular_formula']=='H2O')
]
assert len(the_record)==1
the_record=the_record.iloc[0]
```

```python
new_props=storage['results']['properties']
new_moms=new_props['MolMom_FinFie_through_eng']
new_pols=new_props['MolPol_FinFie_through_eng']
```

```python
pairs=dict(
    dipoles=([new_moms[tag] for tag in ['x','y','z']], the_record['D']),
    quadrupoles=([new_moms[tag] for tag in ['xx','xy','yy','xz','yz','zz']], the_record['Q']),
    ddpols=([new_pols[f"{tag}@{tag}"] for tag in ['x','y','z']], the_record['DDP']),
    qqpols=([new_pols[f"{tag}@{tag}"] for tag in ['xx','xy','yy','xz','yz','zz']], the_record['QQP']),
)
from functools import partial
for k,p in pairs.items():
    print(k)
    print_ind=partial(print," "*4)
    for new,old in zip(*p):
        diff=(new/old-1)*100 if np.abs(old)>1.e-4 else 0
        print_ind(f"{diff:8.4f} {new:8.4f} {old:8.4f}")
        
```
